```python
import matplotlib.pyplot as plt
```

```python
WIDTH = 8

fig, ax1 = plt.subplots()
ax2 = ax1.twinx()

ax1.plot(range(1, 51), [1] * 50, color="red", linewidth=WIDTH)
ax1.plot(range(97, 144), [2] * 47, color="red", linewidth=WIDTH)
ax1.plot(range(51, 106), [3] * 55, color="red", linewidth=WIDTH)
ax1.plot(range(51, 97), [4] * 46, color="red", linewidth=WIDTH)
ax1.plot(range(144, 176), [5] * 32, color="red", linewidth=WIDTH)
ax1.plot(range(119, 176), [6] * 57, color="red", linewidth=WIDTH)
ax1.plot(range(150, 165), [7] * 15, color="red", linewidth=WIDTH)
ax1.plot(range(176, 238), [8] * 62, color="red", linewidth=WIDTH)
ax1.set_ylabel("Numer zadania")

ax2.plot(range(1, 238), [9] * 50 + [15] * 46 + [28] * 9 + [17] * 13 + [24] * 25 + [20] * 6 + [27] * 15 + [20] * 11 + [17] * 62, color="blue")
ax2.set_ylabel("Liczba użytego zasobu")

ax1.set_xlabel("Oś czasu")

fig.savefig("ex4-diagram.pdf", format='pdf')
```
