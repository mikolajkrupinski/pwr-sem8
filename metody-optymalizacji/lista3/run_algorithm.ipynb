```python
import os
import re
import subprocess
```

```python
def run_tests(PATH_IN, PATH_DATA, FILE_OUT):
    with open(FILE_OUT, "w") as file_out:
        file_out.write("instance;opt;approx\n")
        for file in os.listdir(PATH_IN):
            file_data_path = f"{PATH_DATA}{file[:-4]}CPLEX.txt"
            with open(file_data_path, "r") as file_data:
                file_data_content = file_data.read()
                expr = re.search(r"Cmax\s+(\d+)", file_data_content)
                opt = int(expr.group(1))

            file_in_path = os.path.join(PATH_IN, file)
            result = subprocess.run(
                ["julia", "ex1.jl", file_in_path],
                stdout=subprocess.PIPE,
                stderr=subprocess.PIPE,
                text=True
            )
            if result.returncode == 0:
                approx = result.stdout.strip()
            else:
                raise RuntimeError(f"Error running Julia script: {result.stderr}")
            
            file_out.write(f"{file[:-4]};{opt};{approx}\n")
    
```

```python
PATH_IN = "data/instancias1a100/"
PATH_DATA = "data/TXT Cplex 2 horas log U(1,100)/"
FILE_OUT = "data/instancias1a100.csv"
run_tests(PATH_IN, PATH_DATA, FILE_OUT)
```
