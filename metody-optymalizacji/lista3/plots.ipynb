```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
DATA_SETS = ["instancias1a100", "instancias100a120", "instancias100a200", "instanciasde10a100", "Instanciasde1000a1100", "JobsCorre", "MaqCorre"]
DATA_SET_NAMES = ["U(1, 100)", "U(100, 120)", "U(100, 200)", "U(10, 100)", "U(1000, 1100)", "JobsCorre", "MaqCorre"]
PATH = "data"
```

```python
colormap = plt.get_cmap("winter")
for i, data_set in enumerate(DATA_SETS):
    df = pd.read_csv(f"{PATH}/{data_set}.csv", sep=";")
    df["error"] = df["approx"] / df["opt"]
    plt.scatter(range(len(df["instance"])), df["error"], label=DATA_SET_NAMES[i], s=10, color=colormap(i / len(DATA_SETS)))

plt.legend(loc='center left', bbox_to_anchor=(1, 0.5))
plt.grid()
plt.xlabel("Indeks instancji")
plt.ylabel("Błąd aproksymacji")
plt.savefig("error_plot.pdf", bbox_inches='tight', dpi=300)
plt.show()
```

```python
for data_set in DATA_SETS:
    df = pd.read_csv(f"{PATH}/{data_set}.csv", sep=";")
    df["error"] = df["approx"] / df["opt"]
    print(f"(min, avg, max) error {data_set}: ({round(df['error'].min(), 4)}, {round(df['error'].mean(), 4)}, {round(df['error'].max(), 4)})")
```
