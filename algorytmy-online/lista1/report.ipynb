```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
EXPERIMENT_SIZES = [100, 500, 1000, 5000, 10_000, 50_000, 100_000]
STRATEGIES = ["none", "move-to-front", "transpose", "count"]
DISTRIBUTIONS = [
    "uniform",
    "geometric",
    "harmonic",
    "biharmonic",
]
```

```python
df = pd.read_csv("costs.csv", sep=";")

for dist in DISTRIBUTIONS:
    df_dist = df[df["distribution"] == dist]
    for order in STRATEGIES:
        plt.plot(EXPERIMENT_SIZES, df_dist[df_dist["ordering"] == order]["cost"] / EXPERIMENT_SIZES, label=order)
    plt.title(f"distribution = {dist}")
    plt.legend()
    plt.show()
```

```python
df = pd.read_csv("costs.csv", sep=";")

for order in STRATEGIES:
    df_order = df[df["ordering"] == order]
    for dist in DISTRIBUTIONS:
        plt.plot(EXPERIMENT_SIZES, df_order[df_order["distribution"] == dist]["cost"] / EXPERIMENT_SIZES, label=dist)
    plt.title(f"strategy = {order}")
    plt.legend()
    plt.show()
```
