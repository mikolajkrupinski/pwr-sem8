```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
d_values = [16, 32, 64, 128, 256]
p_values = [0.01, 0.02, 0.05, 0.1, 0.2, 0.5]
```

```python
df = pd.read_csv("costs.csv", delimiter=";")

for d in d_values:
    df_d = df[df["D"] == d]
    plt.plot(p_values, df_d["cost"] / 65536, marker="s", label=f"D={d}")

plt.xlabel("p")
plt.ylabel("average cost per request")
plt.legend()
plt.show()
```

```python
df = pd.read_csv("costs.csv", delimiter=";")

for p in p_values:
    df_p = df[df["p"] == p]
    plt.plot(d_values, df_p["cost"] / 65536, marker="s", label=f"p={p}")

plt.xlabel("D")
plt.ylabel("average cost per request")
plt.legend()
plt.show()
```
