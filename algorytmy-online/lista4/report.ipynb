```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("costs.csv", sep=";")

df["cost"] = df["cost"] / 65536

df_cube = df[df["graph"] == "torus"]
df_cube_uniform = df_cube[df_cube["dist"] == "uniform"]

grouped = df_cube_uniform.groupby(["d", "algorithm"])["cost"].mean().unstack()

grouped.loc[[16, 32, 64, 128, 256], ["move-to-min", "coin-flip"]].plot(kind="bar")
plt.grid()
plt.show()
```
