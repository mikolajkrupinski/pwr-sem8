```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("costs.csv", sep=";")
df["cost"] = df["cost"] / 65536

GRAPHS = ["torus", "hypercube"]
DISTS = ["uniform", "harmonic", "biharmonic"]
ALGS = ["move-to-min", "coin-flip"]
D = [16, 32, 64, 128, 256]
```

```python
for graph in GRAPHS:
    for dist in DISTS:
        df_gd = df[(df["graph"] == graph) & (df["dist"] == dist)]
        grouped = df_gd.groupby(["d", "algorithm"])["cost"].mean().unstack()
        grouped.loc[D, ALGS].plot(kind="bar")
        plt.title(f"Graph: {graph}, Distribution: {dist}")
        plt.xlabel("D")
        plt.ylabel("Average cost per request")
        plt.grid()
        plt.show()
```

```python
for graph in GRAPHS:
    for alg in ALGS:
        df_gd = df[(df["graph"] == graph) & (df["algorithm"] == alg)]
        grouped = df_gd.groupby(["d", "dist"])["cost"].mean().unstack()
        grouped.loc[D, DISTS].plot(kind="bar")
        plt.title(f"Graph: {graph}, Algorithm: {alg}")
        plt.xlabel("D")
        plt.ylabel("Average cost per request")
        plt.grid()
        plt.show()
```

```python
for dist in DISTS:
    for alg in ALGS:
        df_gd = df[(df["dist"] == dist) & (df["algorithm"] == alg)]
        grouped = df_gd.groupby(["d", "graph"])["cost"].mean().unstack()
        grouped.loc[D, GRAPHS].plot(kind="bar")
        plt.title(f"Distribution: {dist}, Algorithm: {alg}")
        plt.xlabel("D")
        plt.ylabel("Average cost per request")
        plt.grid()
        plt.show()
```
