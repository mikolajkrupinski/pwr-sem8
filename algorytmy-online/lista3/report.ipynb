```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
STRATEGIES = ["next-fit", "random-fit", "first-fit", "best-fit", "worst-fit"]
DISTRIBUTIONS = [
    "uniform",
    "geometric",
    "harmonic",
    "biharmonic",
]
```

```python
df = pd.read_csv("result.csv", sep=";")

# Create a bar plot using matplotlib
fig, ax = plt.subplots(figsize=(10, 6))

# Group data by strategy and distribution, then calculate mean concurrency
grouped = df.groupby(["strategy", "distribution"])["concurrency"].mean().unstack()

# Plot the data
grouped.loc[STRATEGIES, DISTRIBUTIONS].plot(kind="bar", ax=ax)

# Set plot labels and title
ax.set_xlabel("Strategy")
ax.set_ylabel("Concurrency")
ax.legend(title="Distribution")
plt.ylim(1.0, 1.32)
plt.grid()

# Show the plot
plt.show()
```
