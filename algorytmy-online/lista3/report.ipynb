```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
STRATEGIES = ["next-fit", "random-fit", "first-fit", "best-fit", "worst-fit"]
DISTRIBUTIONS = [
    "uniform",
    "geometric",
    "harmonic",
    "biharmonic",
]
```

```python
df = pd.read_csv("result.csv", sep=";")

fig, ax = plt.subplots(figsize=(10, 6))

grouped = df.groupby(["strategy", "distribution"])["competitive-ratio"].mean().unstack()

grouped.loc[STRATEGIES, DISTRIBUTIONS].plot(kind="bar", ax=ax)

ax.set_xlabel("Strategy")
ax.set_ylabel("Competitive ratio")
ax.legend(title="Distribution")
plt.ylim(1.0, 1.32)
plt.grid()

plt.show()
```
