```python
import matplotlib.pyplot as plt
import pandas as pd
import os
```

```python
os.system("echo 'n;approx;error' > ex11.csv")
for n in range(10000, 10000001, 10000):
    os.system(f"echo -n '{n};' >> ex11.csv")
    os.system(f"./a.out {n} >> ex11.csv")
```

```python
df = pd.read_csv("ex11.csv", delimiter=";")

plt.plot(df["n"], df["approx"])
plt.xlabel("Number of points")
plt.ylabel("Approximation")
plt.grid()
plt.show()

plt.plot(df["n"], df["error"])
plt.xlabel("Number of points")
plt.ylabel("Relative error")
plt.grid()
plt.show()
```
