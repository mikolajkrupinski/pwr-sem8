```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("ex17.csv", sep=";")
df["std"] = (df["approx"] / df["n"]) ** 0.5

plt.scatter(df["n"], df["std"], color="red", alpha=0.5)
plt.grid()
plt.xlabel("n")
plt.ylabel("std of estimate")
plt.title("estimate vs n")
plt.show()
```
