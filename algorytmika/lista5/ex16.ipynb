```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("ex16-3-3.csv", sep=";")

df_uniform = df[df["method"] == "uniform"]
plt.hist(df_uniform["bucket"], weights=df_uniform["load"], bins=len(df_uniform["bucket"].unique()))
plt.title("UNIFORM PLACEMENT (m=10^3, n=10^3)")
plt.xlabel("Bucket")
plt.ylabel("Load")
plt.show()

df_power = df[df["method"] == "power"]
plt.hist(df_power["bucket"], weights=df_power["load"], bins=len(df_power["bucket"].unique()))
plt.title("POWER OF TWO CHOICES (m=10^3, n=10^3)")
plt.xlabel("Bucket")
plt.ylabel("Load")
plt.show()
```
